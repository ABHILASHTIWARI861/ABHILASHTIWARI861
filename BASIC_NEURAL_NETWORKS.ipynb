```python
import tensorflow as tf
import numpy as np
import pandas as pd
from tensorflow import keras

x=np.array([1,2,3,4,5],dtype=int)
y=np.array([40,50,60,70,80],dtype=int)

# Define the model
model=keras.Sequential([keras.layers.Dense(units=1,input_shape=[1])])

#Compile
model.compile(
    optimizer='adam',
    loss='mean_squared_error'
)

#Train
model.fit(x,y,epochs=100,verbose=0)

#Predict
prediction = model.predict(np.array([6]),verbose=0)

print("Predicted Marks: ",prediction[0][0])
```
