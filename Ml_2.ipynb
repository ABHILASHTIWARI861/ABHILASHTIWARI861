```python
import pandas as pd

# Load dataset
df = pd.read_csv("anxiety.csv", encoding='latin1', sep=';')

# Features and target
features = ['Trust_Name']
target = 'Expected'

# Initial hypothesis
hypothesis = ['Ø'] * len(features)

# Find-S algorithm
for index, row in df.iterrows():

    # Consider only positive examples
    if row[target] == 'Yes':

        # First positive example
        if hypothesis == ['Ø'] * len(features):
            for i in range(len(features)):
                hypothesis[i] = row[features[i]]

        # Generalize hypothesis for next positive examples
        else:
            for i in range(len(features)):
                if hypothesis[i] != row[features[i]]:
                    hypothesis[i] = '?'

    print("After row", index + 1, ":", hypothesis)

print("\nFinal Hypothesis:")
print(hypothesis)
```
