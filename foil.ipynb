```python
  import math

def cover(rule, data):
    return [x for x in data if all(x[k] == v for k, v in rule)]

def foil(data, target):
    pos = [x for x in data if x[target] == "Yes"]
    neg = [x for x in data if x[target] == "No"]
    rules = []

    while pos:
        rule = []

        while cover(rule, neg):
            best, bg = None, -999

            for k in data[0]:
                if k == target:
                    continue

                for v in set(x[k] for x in data):
                    if (k, v) in rule:
                        continue

                    r = rule + [(k, v)]
                    p0, n0 = len(cover(rule, pos)), len(cover(rule, neg))
                    p1, n1 = len(cover(r, pos)), len(cover(r, neg))

                    if p1 and p1 + n1:
                        g = p1 * (
                            math.log2(p1 / (p1+n1)) -
                            math.log2(p0 / (p0+n0))
                        )

                        if g > bg:
                            best, bg = (k, v), g

            if best is None:
                break

            rule.append(best)

        rules.append(rule)
        pos = [x for x in pos if x not in cover(rule, pos)]

    return rules


# Example
data = [
    {"Outlook":"Sunny", "Humidity":"High", "Play":"No"},
    {"Outlook":"Sunny", "Humidity":"Normal", "Play":"Yes"},
    {"Outlook":"Rain", "Humidity":"High", "Play":"Yes"},
    {"Outlook":"Rain", "Humidity":"Normal", "Play":"Yes"}
]

rules = foil(data, "Play")

for r in rules:
    print(r)
```
